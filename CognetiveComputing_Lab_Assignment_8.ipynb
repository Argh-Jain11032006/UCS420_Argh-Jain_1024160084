## Q1. Text Preprocessing and Tokenization

**Aim:** To perform text preprocessing using Python and NLTK, including lowercase conversion, punctuation removal, word and sentence tokenization, stopword removal, and word frequency distribution.


```python
# Install the required NLTK library
!pip -q install nltk

# Import required libraries
import re
import nltk
import matplotlib.pyplot as plt

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download the required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

print("Libraries imported successfully!")
```

```python
# Q1: Original paragraph (6 sentences)

text = """
I love reading books because they expand my imagination and knowledge.
Technology has changed the way people read and learn.
Digital libraries allow students to access thousands of books online.
Artificial intelligence can recommend books based on personal interests.
Reading regularly improves vocabulary, creativity, and critical thinking.
I believe books and technology can work together to make learning better.
"""

print("Original Paragraph:\n")
print(text)
```

```python
# Convert the text to lowercase
lower_text = text.lower()

# Remove punctuation using regular expressions
clean_text = re.sub(r'[^\w\s]', '', lower_text)

print("Cleaned Text:\n")
print(clean_text)
```

```python
# Sentence tokenization
sentences = sent_tokenize(text)

# Word tokenization
words = word_tokenize(clean_text)

print("Sentence Tokens:")
for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")

print("\nWord Tokens:")
print(words)

print("\nTotal sentences:", len(sentences))
print("Total word tokens:", len(words))
```

```python
# Python's built-in split()
split_words = clean_text.split()

# NLTK word tokenization
nltk_words = word_tokenize(clean_text)

print("Python split() output:")
print(split_words)

print("\nNLTK word_tokenize() output:")
print(nltk_words)

print("\nNumber of tokens using split():", len(split_words))
print("Number of tokens using word_tokenize():", len(nltk_words))
```

```python
# Load English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords
filtered_words = [
    word for word in nltk_words
    if word not in stop_words and word.isalpha()
]

print("Words after stopword removal:")
print(filtered_words)

print("\nTotal words before filtering:", len(nltk_words))
print("Total words after filtering:", len(filtered_words))
```

```python
# Calculate word frequencies
frequency_distribution = FreqDist(filtered_words)

print("Word Frequency Distribution:\n")
for word, frequency in frequency_distribution.most_common():
    print(f"{word}: {frequency}")

# Plot the 10 most frequent words
plt.figure(figsize=(10, 5))
frequency_distribution.plot(10, title="Top 10 Most Frequent Words")
plt.show()
```

## Conclusion

Text preprocessing was successfully performed using Python and NLTK. The paragraph was converted to lowercase, punctuation was removed using regular expressions, and words and sentences were tokenized. The outputs of Python's `split()` and NLTK's `word_tokenize()` were compared. Stopwords were removed, and the frequency distribution of the remaining words was calculated and visualized. These preprocessing techniques help prepare textual data for further Natural Language Processing tasks.
