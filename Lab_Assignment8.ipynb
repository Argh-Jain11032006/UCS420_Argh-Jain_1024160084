```python
!pip -q install nltk textblob wordcloud scikit-learn tensorflow

import re
import random
import numpy as np
import nltk
import matplotlib.pyplot as plt

from collections import Counter
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.probability import FreqDist
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from textblob import TextBlob
from wordcloud import WordCloud

import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.utils import to_categorical

# Download required NLTK resources
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

# Set random seeds for reproducibility
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

print("All libraries and resources are ready!")


# ============================================================
# Q1: TEXT PREPROCESSING AND TOKENIZATION
# ============================================================

print("\n" + "=" * 65)
print("Q1: TEXT PREPROCESSING AND TOKENIZATION")
print("=" * 65)

# Original paragraph: Books and Technology
text = """
I love reading books because they expand my imagination and knowledge.
Technology has changed the way people read and learn.
Digital libraries allow students to access thousands of books online.
Artificial intelligence can recommend books based on personal interests.
Reading regularly improves vocabulary, creativity, and critical thinking.
I believe books and technology can work together to make learning better.
"""

print("\nOriginal Paragraph:\n", text)

# 1. Convert to lowercase and remove punctuation
lower_text = text.lower()
clean_text = re.sub(r'[^\w\s]', '', lower_text)

print("\nCleaned Text:\n", clean_text)

# 2. Tokenize into sentences and words
sentences = sent_tokenize(text)
words = word_tokenize(clean_text)

print("\nSentence Tokens:")
for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")

print("\nWord Tokens:")
print(words)

# 3. Compare split() and word_tokenize()
split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("\nPython split() Output:")
print(split_words)

print("\nNLTK word_tokenize() Output:")
print(nltk_words)

print("\nNumber of tokens using split():", len(split_words))
print("Number of tokens using word_tokenize():", len(nltk_words))

print("\nComparison:")
print("split() divides text at whitespace by default.")
print("word_tokenize() uses NLP tokenization rules.")
print("Punctuation was removed before both methods were applied.")

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in nltk_words
    if word.isalpha() and word not in stop_words
]

print("\nWords after Stopword Removal:")
print(filtered_words)

# 5. Word frequency distribution
freq_dist = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, frequency in freq_dist.most_common():
    print(f"{word}: {frequency}")

plt.figure(figsize=(10, 5))
freq_dist.plot(10, title="Q1: Top 10 Most Frequent Words")
plt.show()


# ============================================================
# Q2: STEMMING AND LEMMATIZATION
# ============================================================

print("\n" + "=" * 65)
print("Q2: STEMMING AND LEMMATIZATION")
print("=" * 65)

# 1. Extract alphabetic words using re.findall()
words_q2 = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("\nAlphabetic Words:")
print(words_q2)

# 2. Remove stopwords
filtered_q2 = [
    word for word in words_q2
    if word not in stop_words
]

print("\nWords after Stopword Removal:")
print(filtered_q2)

# 3. Perform stemming
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_q2
]

print("\nStemmed Words:")
print(stemmed_words)

# 4. Perform lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_q2
]

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Compare outputs
print("\nStemming vs Lemmatization:")
print(f"{'Original':<18}{'Stemmed':<18}{'Lemmatized':<18}")
print("-" * 54)

for original, stem, lemma in zip(
    filtered_q2, stemmed_words, lemmatized_words
):
    print(f"{original:<18}{stem:<18}{lemma:<18}")

print("\nInterpretation:")
print("Stemming reduces words using word-reduction rules.")
print("Lemmatization attempts to produce dictionary forms.")
print("Stemming is useful for speed; lemmatization for meaningful forms.")


# ============================================================
# Q3: BAG OF WORDS AND TF-IDF
# ============================================================

print("\n" + "=" * 65)
print("Q3: BAG OF WORDS AND TF-IDF")
print("=" * 65)

# Three original short texts
texts_q3 = [
    "Artificial intelligence improves healthcare through smart diagnosis.",
    "Machine learning helps computers discover patterns from data.",
    "Healthcare technology uses artificial intelligence for medical research."
]

print("\nInput Texts:")
for i, sentence in enumerate(texts_q3, 1):
    print(f"Text {i}: {sentence}")

# 1. Bag of Words
bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(texts_q3)

print("\nBag of Words Vocabulary:")
print(bow_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

# 2. TF-IDF
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts_q3)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTF-IDF Matrix:")
print(np.round(tfidf_matrix.toarray(), 3))

# 3. Top 3 keywords for each text
print("\nTop 3 Keywords for Each Text:")

for i, row in enumerate(tfidf_matrix.toarray()):
    top_indices = np.argsort(row)[::-1][:3]

    print(f"\nText {i + 1}:")
    for idx in top_indices:
        print(f"{feature_names[idx]}: {row[idx]:.4f}")

print("\nInterpretation:")
print("Bag of Words records word counts.")
print("TF-IDF assigns greater importance to relatively distinctive words.")
print("Higher TF-IDF scores indicate words that are more informative")
print("within a text relative to the other texts in this collection.")


# ============================================================
# Q4: JACCARD AND COSINE SIMILARITY
# ============================================================

print("\n" + "=" * 65)
print("Q4: JACCARD AND COSINE SIMILARITY")
print("=" * 65)

# Two technology descriptions
text1 = """
Artificial intelligence enables computers to learn from data.
AI systems recognize patterns and automate complex tasks.
Machine learning improves predictions using training examples.
"""

text2 = """
Blockchain stores information in a distributed digital ledger.
Blockchain technology records transactions securely across computers.
Cryptographic methods help protect the integrity of stored records.
"""

# Preprocess and tokenize
def preprocess_for_similarity(text):
    text = text.lower()
    return re.findall(r'\b[a-zA-Z]+\b', text)

tokens1 = preprocess_for_similarity(text1)
tokens2 = preprocess_for_similarity(text2)

print("\nTechnology 1: Artificial Intelligence")
print(text1)

print("Technology 2: Blockchain")
print(text2)

print("\nTokens for Technology 1:")
print(tokens1)

print("\nTokens for Technology 2:")
print(tokens2)

# 1. Jaccard Similarity
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union) if union else 0

print("\nCommon Words:")
print(intersection)

print("Jaccard Similarity:", round(jaccard, 4))

# 2. Cosine Similarity using TF-IDF
similarity_vectorizer = TfidfVectorizer()
similarity_matrix = similarity_vectorizer.fit_transform(
    [text1, text2]
)

cosine_score = cosine_similarity(
    similarity_matrix[0:1],
    similarity_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(float(cosine_score), 4))

# 3. Compare metrics
print("\nInterpretation:")
print("Jaccard compares the overlap between unique word sets.")
print("Cosine similarity compares the directions of TF-IDF vectors.")
print("Both metrics help compare texts, but measure similarity differently.")

if jaccard > cosine_score:
    print("Jaccard similarity is numerically higher for this example.")
elif cosine_score > jaccard:
    print("Cosine similarity is numerically higher for this example.")
else:
    print("Both similarity scores are equal for this example.")

print("For these technology descriptions, both scores may be low because")
print("the texts use different technical vocabulary.")


# ============================================================
# Q5: SENTIMENT ANALYSIS AND WORD CLOUD
# ============================================================

print("\n" + "=" * 65)
print("Q5: SENTIMENT ANALYSIS AND WORD CLOUD")
print("=" * 65)

# Sample product/service reviews
reviews = [
    "This laptop is excellent, fast, and very reliable.",
    "The product is amazing and offers excellent performance.",
    "The service was terrible and the product stopped working.",
    "Very disappointing quality and poor customer support.",
    "The product is okay and works as expected.",
    "The service was average with nothing special."
]

# Function to classify sentiment
def classify_sentiment(polarity):
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

# 1. Polarity and subjectivity
positive_reviews = []

print("\nSentiment Analysis Results:")

for i, review in enumerate(reviews, 1):
    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity
    sentiment = classify_sentiment(polarity)

    print(f"\nReview {i}: {review}")
    print(f"Polarity: {polarity:.3f}")
    print(f"Subjectivity: {subjectivity:.3f}")
    print(f"Sentiment: {sentiment}")

    if sentiment == "Positive":
        positive_reviews.append(review)

# 2. Sentiment counts
sentiments = [
    classify_sentiment(TextBlob(review).sentiment.polarity)
    for review in reviews
]

sentiment_counts = Counter(sentiments)

print("\nSentiment Summary:")
for sentiment in ["Positive", "Negative", "Neutral"]:
    print(f"{sentiment}: {sentiment_counts[sentiment]}")

# Plot sentiment counts
plt.figure(figsize=(7, 4))
plt.bar(
    sentiment_counts.keys(),
    sentiment_counts.values()
)
plt.title("Q5: Sentiment Classification")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.show()

# 3. Word cloud using positive reviews
positive_text = " ".join(positive_reviews)

if positive_text.strip():
    wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white"
    ).generate(positive_text)

    plt.figure(figsize=(12, 6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Q5: Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews found; word cloud cannot be generated.")

print("\nInterpretation:")
print("Polarity ranges from -1 to +1.")
print("Negative values indicate negative sentiment.")
print("Positive values indicate positive sentiment.")
print("Subjectivity ranges from 0 (objective) to 1 (subjective).")
print("TextBlob results are estimates and may not understand every context.")


# ============================================================
# Q6: TEXT GENERATION USING AN LSTM MODEL
# ============================================================

print("\n" + "=" * 65)
print("Q6: TEXT GENERATION USING LSTM")
print("=" * 65)

# Approximately 100 words of training data
training_text = """
Books open doors to knowledge and imagination.
Reading helps students develop creativity and critical thinking.
Technology makes learning accessible through digital libraries.
Students can explore science mathematics history and literature online.
Artificial intelligence helps learners discover useful information.
Machine learning identifies patterns and supports intelligent applications.
Good education combines books technology curiosity and practice.
Reading regularly improves vocabulary communication and understanding.
Students learn better when they ask questions and solve problems.
Knowledge grows when people share ideas and explore new possibilities.
Digital tools help learners organize information and study efficiently.
Creative thinking allows students to develop innovative solutions.
Books and technology together create opportunities for lifelong learning.
Every student can improve through patience effort and consistent practice.
"""

# 1. Tokenize text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])

total_words = len(tokenizer.word_index) + 1

tokens = tokenizer.texts_to_sequences([training_text])[0]

print("\nVocabulary Size:", total_words - 1)
print("Tokenized Training Text:")
print(tokens)

# 2. Create input sequences
sequences = []

for i in range(1, len(tokens)):
    sequence = tokens[:i + 1]
    sequences.append(sequence)

# Pad sequences to equal length
max_sequence_length = max(len(seq) for seq in sequences)

padded_sequences = pad_sequences(
    sequences,
    maxlen=max_sequence_length,
    padding='pre'
)

padded_sequences = np.array(padded_sequences)

# Split inputs and target words
X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]

# One-hot encode target words
y = to_categorical(y, num_classes=total_words)

print("\nNumber of Input Sequences:", len(X))
print("Input Shape:", X.shape)
print("Target Shape:", y.shape)

# 3. Build a simple LSTM model
model = Sequential([
    Embedding(
        input_dim=total_words,
        output_dim=32,
        input_length=X.shape[1]
    ),
    LSTM(64),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

print("\nLSTM Model Summary:")
model.summary()

# 4. Train the model
print("\nTraining the LSTM model...")

history = model.fit(
    X,
    y,
    epochs=100,
    verbose=0
)

print("Training completed!")

print("Final Training Loss:", round(history.history['loss'][-1], 4))
print("Final Training Accuracy:",
      round(history.history['accuracy'][-1], 4))

# Plot training loss
plt.figure(figsize=(8, 4))
plt.plot(history.history['loss'])
plt.title("Q6: LSTM Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

# 5. Generate new text from a seed word
def generate_text(seed_text, next_words=10):
    generated_text = seed_text.lower()

    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences(
            [generated_text]
        )[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=X.shape[1],
            padding='pre',
            truncating='pre'
        )

        predictions = model.predict(token_list, verbose=0)[0]

        # Select the most probable next word
        predicted_index = int(np.argmax(predictions))

        if predicted_index == 0:
            break

        predicted_word = tokenizer.index_word.get(predicted_index)

        if predicted_word is None:
            break

        generated_text += " " + predicted_word

    return generated_text

print("\nGenerated Text:")

# Generate three short continuations
seed_words = ["books", "technology", "students"]

for seed in seed_words:
    print(f"\nSeed: {seed}")
    print(generate_text(seed, next_words=8))

print("\nInterpretation:")
print("The LSTM learns word sequences from the training paragraph.")
print("It predicts a likely next word based on the preceding words.")
print("The generated text is experimental because the training dataset")
print("is small and the model has limited linguistic knowledge.")


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("ASSIGNMENT 8 COMPLETED")
print("=" * 65)

print("""
Q1: Text preprocessing, tokenization and word frequency
Q2: Stemming and lemmatization
Q3: Bag of Words and TF-IDF keyword extraction
Q4: Jaccard and cosine similarity
Q5: Sentiment analysis and positive-review word cloud
Q6: LSTM-based text generation

All six questions have been executed.
""")

```
