```python
# Q1. Sensor readings

import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

print("Original temperatures:", temperature)
```

## Q1(a) Add +2°C using vectorization

```python
temperature_corrected = temperature + 2

print("Corrected temperatures (°C):", temperature_corrected)
```

## Q1(b) Convert Celsius to Fahrenheit

```python
fahrenheit = (9/5) * temperature_corrected + 32

print("Temperatures in Fahrenheit:", fahrenheit)
```

## Q1(c) Identify readings greater than 32°C

```python
readings_greater_32 = temperature[temperature > 32]

print("Original readings greater than 32°C:", readings_greater_32)
```

## Q1(d) Count readings greater than 32°C

```python
count_greater_32 = np.sum(temperature > 32)

print("Number of readings greater than 32°C:", count_greater_32)
```

## Q1(e) Explanation

**Vectorization** applies an operation to an entire NumPy array at once rather than explicitly processing
each element with a Python `for` loop. NumPy performs these operations efficiently using optimized
low-level routines.

**Boolean indexing** creates a Boolean mask such as `temperature > 32` and directly selects the values
that satisfy the condition. This makes filtering concise and generally faster than manually checking
every element in a Python loop.


## Q2. Daily steps

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix:")
print(steps)
```

### Q2(a) Total steps recorded

```python
total_steps = np.sum(steps)
print("Total steps:", total_steps)
```

### Q2(b) Mean number of steps

```python
mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)
```

### Q2(c) Maximum and minimum values

```python
max_steps = np.max(steps)
min_steps = np.min(steps)

print("Maximum steps:", max_steps)
print("Minimum steps:", min_steps)
```

### Q2(d) Total steps for each day using `axis=0`

```python
total_each_day = np.sum(steps, axis=0)

print("Total steps for each day:", total_each_day)
```

### Q2(e) Total steps for each user using `axis=1`

```python
total_each_user = np.sum(steps, axis=1)

print("Total steps for each user:", total_each_user)
```

### Q2(f) Position of maximum value using `argmax()`

```python
flat_index = np.argmax(steps)
max_position = np.unravel_index(flat_index, steps.shape)

print("Maximum value:", steps[max_position])
print("User/day position (0-based index):", max_position)
print("User number (1-based):", max_position[0] + 1)
print("Day number (1-based):", max_position[1] + 1)
```

## Q3. Array slicing, copying, reshaping and flattening

```python
# Q3(a)
original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)
```

### Q3(b–c) Slice and modify it

```python
subset = original[1:4]
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

# NumPy slices are views by default, so the original array is affected.
```

### Q3(d) Create a copied slice and modify it

```python
copied = original[1:4].copy()
copied[0] = 500

print("Original after modifying copied array:", original)
print("Copied array:", copied)

# copy() creates an independent array, so the original is not changed by this modification.
```

### Q3(e) Create a 3 × 4 matrix

```python
matrix = np.arange(1, 13).reshape(3, 4)

print("3 × 4 matrix:")
print(matrix)
```

### Q3(f) Indexing and slicing

```python
first_row = matrix[0, :]
last_row = matrix[-1, :]
second_column = matrix[:, 1]
rows_1_to_2_cols_2_to_3 = matrix[1:3, 2:4]

print("First row:", first_row)
print("Last row:", last_row)
print("Second column:", second_column)
print("Rows 1–2 and columns 2–3 (1-based):")
print(rows_1_to_2_cols_2_to_3)
```

### Q3(g–i) `flatten()` vs `ravel()`

```python
flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("flatten():", flat_array)
print("ravel():", ravel_array)

# Modify ravel() output
ravel_array[0] = 1000
print("\nAfter modifying ravel()[0] = 1000")
print("Ravel array:", ravel_array)
print("Original matrix:")
print(matrix)

# Reset for a clean demonstration
matrix = np.arange(1, 13).reshape(3, 4)

# Modify flatten() output
flat_array = matrix.flatten()
flat_array[0] = 2000
print("\nAfter modifying flatten()[0] = 2000")
print("Flattened array:", flat_array)
print("Original matrix:")
print(matrix)

# ravel() normally returns a view when possible, while flatten() always returns a copy.
```

### Q3(j) Shape, dimensions, size and data type

```python
print("Shape:", matrix.shape)
print("Number of dimensions (ndim):", matrix.ndim)
print("Number of elements (size):", matrix.size)
print("Data type (dtype):", matrix.dtype)
```

## Q4. Cognitive-assistive system — Ordinary Least Squares

We use the given feature matrix:

- Column 1: Sleep hours
- Column 2: Activity level
- Column 3: Stress level

and target vector `y`, the assistance score.


```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("X =")
print(X)
print("\ny =", y)
```

### Q4(a) Shape and dimensions

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

### Q4(b) Transpose `X.T`

```python
XT = X.T

print("X.T =")
print(XT)
```

**Interpretation:** `X.T` changes the matrix from 4 rows × 3 columns to 3 rows × 4 columns.
Each original feature becomes a row, containing that feature's values for all four users.


### Q4(c) Matrix product `X.T @ X`

```python
XTX = X.T @ X

print("X.T @ X =")
print(XTX)
```

### Q4(d–e) Matrix inverse and OLS coefficients

```python
XTX_inv = np.linalg.inv(XTX)

beta = XTX_inv @ X.T @ y

print("(X.T @ X)^(-1) =")
print(XTX_inv)

print("\nOLS coefficients beta:")
print(beta)
```

### Q4(f) Interpretation of coefficients

The OLS model is:

`assistance_score = β1(sleep) + β2(activity) + β3(stress)`

Each coefficient estimates the change in predicted assistance score associated with a one-unit increase
in that feature while the other two features are held constant.

Because the assignment specifies this exact equation and does **not** include an intercept column, the
calculated coefficients correspond directly to the three supplied features.


### Q4(g) Predict the score for a new user

```python
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("New user:", new_user)
print("Predicted assistance score:", predicted_score)
```
